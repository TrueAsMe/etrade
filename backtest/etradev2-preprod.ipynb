```python
# data_feed.py — MYM Data Layer (Databento Historical + Live)
# ---------------------------------------------------------------
# Databento ohlcv-1m -> _clean_1m -> resample_ohlc -> 5m / 15m / 1h.
#   Historical  : deep backfill of the rolling window (batch snapshot).
#   Live        : replays the gap since the last backfilled bar, then
#                 streams new 1m bars, keeping frames up to the minute.
#
# Public entry points:
#   load_all_timeframes()        -> (df_1m, df_5m, df_15m, df_1h)   # batch, signals.py
#   run_live_feed(on_new_bar=..) -> async: backfill then stream     # on_new_bar((1m,5m,15m,1h))
#   estimate_databento_cost()    -> float  (call before a backfill; spends NO credits)
#
# IMPORT-SAFE: importing this module performs no network calls and spends
# no Databento credits. Every pull is behind a function you call explicitly.
# ---------------------------------------------------------------

from __future__ import annotations

import os
import asyncio
import pandas as pd
import databento as db
import sys
from pathlib import Path
from dotenv import load_dotenv
## Give the notebook a .env to find
load_dotenv(os.path.expanduser("~/prod/etrade/.env"))
sys.path.insert(0, os.path.expanduser("~/prod/etrade"))   # where config.py lives
import config

# -- CONTRACT / DATASET ----------------------------------------
DATASET  = "GLBX.MDP3"                              # CME Globex (covers CBOT / MYM)
# Explicit contract, pinned manually to avoid continuous-roll splicing.
# MYMU6 = Micro E-mini Dow, SEPTEMBER 2026 (root+month+1-digit year; U=Sep, 6=2026).
# >>> UPDATE at each quarterly roll (H=Mar, M=Jun, U=Sep, Z=Dec) <<<
SYMBOL   = "MYMU6"
STYPE_IN = "raw_symbol"                             # explicit contract symbol
# --- continuous layout (use INSTEAD of the two lines above) ---
#SYMBOL   = "MYM.v.0"                               # volume-weighted continuous
#STYPE_IN = "continuous"
SCHEMA   = "ohlcv-1m"                               # 5m/15m/1h are resampled from this
DATA_LAG = pd.Timedelta(minutes=16)    # keep Historical end inside the available edge
PX_SCALE = 1e9                                      # Live prices are raw ints; Historical .to_df() is pre-scaled

# -- CREDENTIALS -----------------------------------------------
def _client():
    key = getattr(config, "DATABENTO_API_KEY", "") or os.getenv("DATABENTO_API_KEY", "")
    if not key:
        raise RuntimeError("DATABENTO_API_KEY not set (config.py or .env).")
    return key

# -- CLEANING / RESAMPLE (shared by Historical + Live) ---------
def _clean_1m(df, label="1m"):
    """Raw Databento ohlcv-1m -> tz-aware UTC 'datetime' index, spreads & dupes
    removed, OHLC only — the SAME cleaning the backtest CSV got, so live and
    backtest bars are treated identically."""
    if df.empty:
        print(f"{label:6} |      0 bars | (no data returned)")
        return pd.DataFrame(columns=["open", "high", "low", "close"])
    df = df.copy()
    df.index = pd.to_datetime(df.index, utc=True)
    df.index.name = "datetime"
    df = df.dropna().sort_index()
    if "symbol" in df.columns:                                       # drop roll-spread rows
        df = df[~df["symbol"].astype(str).str.contains("-")]
    dupes = int(df.index.duplicated().sum())
    df = df[~df.index.duplicated(keep="last")]
    df = df[["open", "high", "low", "close"]].copy()
    d0, d1 = df.index[0].date(), df.index[-1].date()
    days = (d1 - d0).days
    print(f"{label:6} | {len(df):>6} bars | {d0} to {d1} | "
          f"range {df['low'].min():.0f}-{df['high'].max():.0f} | "
          f"{days} days | {days // 7} weeks | dupes removed: {dupes}")
    return df

def resample_ohlc(df, rule):
    return df.resample(rule).agg(
        {"open": "first", "high": "max", "low": "min", "close": "last"}
    ).dropna()

# -- HISTORICAL PULL -------------------------------------------
def _safe_end(end=None):
    """End-time capped safely in the past (Historical API rejects the recent edge)."""
    return end if end is not None else pd.Timestamp.now(tz="UTC") - DATA_LAG

def _pull_1m(start, end=None):
    """Low-level Databento ohlcv-1m pull -> cleaned 1m DataFrame."""
    store = db.Historical(_client()).timeseries.get_range(
        dataset=DATASET, symbols=[SYMBOL], schema=SCHEMA, stype_in=STYPE_IN,
        start=start, end=_safe_end(end),
    )
    return _clean_1m(store.to_df())

def load_all_timeframes(days=config.DATA_DAYS, cache_path=None):
    """Batch snapshot for signals.py. Returns (df_1m, df_5m, df_15m, df_1h).

    cache_path=None : fresh pull of the full window (simplest).
    cache_path set  : incremental — read cache, pull only the gap since the last
                      cached minute, append, dedupe, trim, rewrite. Saves credits.
                      Use a DEDICATED cache file, not the backtest base CSV.
    """
    cutoff = pd.Timestamp.now(tz="UTC") - pd.Timedelta(days=days)

    if cache_path and os.path.exists(cache_path):
        cached = pd.read_csv(cache_path)
        cached["datetime"] = pd.to_datetime(cached["datetime"], utc=True)
        cached = cached.set_index("datetime").sort_index()
        fresh = _pull_1m(start=cached.index.max())                  # only the gap
        df_1m = pd.concat([cached[["open", "high", "low", "close"]], fresh])
        df_1m = df_1m[~df_1m.index.duplicated(keep="last")].sort_index()
    else:
        df_1m = _pull_1m(start=cutoff)

    df_1m = df_1m[df_1m.index >= cutoff]                            # trim to rolling window
    if cache_path:
        df_1m.reset_index().to_csv(cache_path, index=False)

    df_5m  = resample_ohlc(df_1m, "5min")
    df_15m = resample_ohlc(df_1m, "15min")
    df_1h  = resample_ohlc(df_1m, "1h")
    print(f"OK Databento | 1m:{len(df_1m)} 5m:{len(df_5m)} 15m:{len(df_15m)} 1h:{len(df_1h)}")
    return df_1m, df_5m, df_15m, df_1h

def estimate_databento_cost(days=config.DATA_DAYS):
    """$ cost of the `days`-day 1m pull, WITHOUT spending credits."""
    start = pd.Timestamp.now(tz="UTC") - pd.Timedelta(days=days)
    return db.Historical(_client()).metadata.get_cost(
        dataset=DATASET, symbols=[SYMBOL], schema=SCHEMA, stype_in=STYPE_IN,
        start=start, end=_safe_end(),
    )

# -- LIVE STREAM -----------------------------------------------
async def run_live_feed(days=config.DATA_DAYS, on_new_bar=None):
    """Historical backfill, then live 1m stream. Fires
    on_new_bar((df_1m, df_5m, df_15m, df_1h)) on each closed minute.
    All frame mutation stays on the asyncio loop thread."""
    loop, queue = asyncio.get_running_loop(), asyncio.Queue()
    window = pd.Timedelta(days=days)

    df_1m = _pull_1m(start=pd.Timestamp.now(tz="UTC") - window)      # seed
    start_ns = int(df_1m.index.max().value) + 1 if not df_1m.empty else None  # gapless seam
    print(f"OK backfill | 1m: {len(df_1m)} | live replay from ns={start_ns}")

    def enqueue(rec):                                               # runs on Databento's thread
        loop.call_soon_threadsafe(queue.put_nowait, rec)

    async def consume():
        nonlocal df_1m
        while True:
            rec = await queue.get()
            if isinstance(rec, db.OHLCVMsg):                        # ignore symbol-map/system/error msgs
                ts = pd.Timestamp(rec.ts_event, unit="ns", tz="UTC")
                df_1m.loc[ts] = [rec.open / PX_SCALE, rec.high / PX_SCALE,
                                 rec.low / PX_SCALE, rec.close / PX_SCALE]
                df_1m = df_1m[~df_1m.index.duplicated(keep="last")].sort_index()
                df_1m = df_1m[df_1m.index >= df_1m.index.max() - window]
                if on_new_bar:
                    on_new_bar((df_1m.copy(), resample_ohlc(df_1m, "5min"),
                                resample_ohlc(df_1m, "15min"), resample_ohlc(df_1m, "1h")))
            queue.task_done()

    consumer = asyncio.create_task(consume())

    live = db.Live(key=_client())
    live.subscribe(dataset=DATASET, schema=SCHEMA, stype_in=STYPE_IN, symbols=[SYMBOL], start=start_ns)
    live.add_callback(enqueue)
    await loop.run_in_executor(None, live.start)
    print(f"OK live | streaming {SYMBOL}")

    try:
        while True:
            await asyncio.sleep(1)
    except asyncio.CancelledError:
        live.stop()
        consumer.cancel()
        raise

if __name__ == "__main__":
    # Manual smoke test only — never runs on import.
    print(f"Est. cost ({config.DATA_DAYS}d 1m): ${estimate_databento_cost():.4f}")
    df_1m, df_5m, df_15m, df_1h = load_all_timeframes()
```

```python
# data_feed.py — MYM Data Layer (Historical API || Databento || For Backtesting)
# Cell 1 — Loading the Data (backtest: Databento -> 1m -> resample)
# ---------------------------------------------------------------
# Single source of market data for the strategy.
#   Databento ohlcv-1m  ->  _clean  ->  resample_ohlc  ->  5m / 15m /1h
# 1-minute is the only timeframe pulled; everything else is resampled
# so the source stays consistent with the backtest base.
#
# Public entry point (unchanged contract used by signals.py):
#   load_all_timeframes()  ->  (df_1m, df_5m, df_15m, df_1h)
#
# No live connection is required — Databento Historical serves bars up
# to the latest available minute (a few minutes behind real time), which
# is plenty for a 5m/15m strategy. 
# 
# Schwab is used ONLY for live bracket
# order testing (see schwab_data.py), not for data.
# ---------------------------------------------------------------

from __future__ import annotations

import os
import pandas as pd
import sys
from pathlib import Path
from dotenv import load_dotenv

## Give the notebook a .env to find
load_dotenv(os.path.expanduser("~/prod/etrade/.env"))
sys.path.insert(0, os.path.expanduser("~/prod/etrade"))   # where config.py lives
import config

# -- DATABENTO CONTRACT/DATASET --------------------------------
DATASET     = "GLBX.MDP3"     # CME Globex (covers CBOT, where MYM lists)
# Explicit contract, pinned manually for now to avoid continuous-roll splicing.
# MYMU6 = Micro E-mini Dow, SEPTEMBER 2026 (CME format: root+month+1-digit year;
# U=Sep, 6=2026). Matches TradersPost MYMU2026 / Schwab /MYMU26.
# >>> UPDATE THIS at each quarterly roll (H=Mar, M=Jun, U=Sep, Z=Dec) <<<
#SYMBOL     = "MYMU6"
SYMBOL      = "MYM.v.0"
SCHEMA      = "ohlcv-1m"      # 1-minute OHLCV; 5m/15m are resampled from this
STYPE_IN    = "continuous"    # explicit native contract symbol (not a continuous code) "raw_symbol"

# Databento Historical for this plan/dataset lags real time by ~8 hours; requesting
# closer to "now" returns 422 'dataset_unavailable_range'. Cap the request end this
# far back (8h + 30m margin to stay safely inside the available window). Lower toward
# 8h for slightly fresher data, or shrink it with a live CME subscription.
DATA_LAG = pd.Timedelta(hours=1)#hours=4,

def _client():
    """Databento Historical client. Key from config.DATABENTO_API_KEY or env."""
    key = getattr(config, "DATABENTO_API_KEY", "") or os.getenv("DATABENTO_API_KEY", "")
    if not key:
        raise RuntimeError("DATABENTO_API_KEY not set (add it to config.py or your .env).")
    import databento as db
    return db.Historical(key)

# -- 1m CLEANING (mirrors the old load_tv_csv treatment) -------
def _clean_1m(df, label="1m"):
    """Raw Databento ohlcv-1m -> tz-aware UTC 'datetime' index, spreads & dupes
    removed, OHLC only -- the SAME cleaning the backtest CSV got, so live and
    backtest bars are treated identically."""
    df = df.copy()
    if df.empty:
        print(f"{label:6} |      0 bars | (no data returned)")
        return pd.DataFrame(columns=["open", "high", "low", "close"])
    df.index = pd.to_datetime(df.index, utc=True)        # ts_event -> tz-aware UTC
    df.index.name = "datetime"
    df = df.dropna().sort_index()

    if "symbol" in df.columns:                            # drop contract-roll spread rows
        df = df[~df["symbol"].astype(str).str.contains("-")]

    dupes = int(df.index.duplicated().sum())              # keep last tick for the minute
    df = df[~df.index.duplicated(keep="last")]
    df = df[["open", "high", "low", "close"]].copy()

    start_date    = df.index[0].date()
    end_date      = df.index[-1].date()
    range_day_cnt = (end_date - start_date).days
    print(f"{label:6} | {len(df):>6} bars | {start_date} to {end_date} | "
          f"Price range: {df['low'].min():.0f} -- {df['high'].max():.0f} | "
          f"{range_day_cnt} days | {int(range_day_cnt / 7)} weeks | dupes removed: {dupes}")
    return df

def _safe_end(end=None):
    """End-time capped safely in the past (Historical API rejects the recent edge)."""
    if end is not None:
        return end
    return pd.Timestamp.now(tz="UTC") - DATA_LAG


def _pull_1m(start, end=None):
    """Low-level Databento ohlcv-1m pull -> cleaned 1m DataFrame (end=None -> now-DATA_LAG)."""
    store = _client().timeseries.get_range(
        dataset=DATASET, symbols=[SYMBOL], schema=SCHEMA, stype_in=STYPE_IN,
        start=start, end=_safe_end(end),
    )
    return _clean_1m(store.to_df())

# -- RESAMPLE HELPER (retained verbatim) -----------------------
def resample_ohlc(df, rule):
    df_resampled = df.resample(rule).agg({
        "open"  : "first",
        "high"  : "max",
        "low"   : "min",
        "close" : "last",
    }).dropna()

    start_date     = df_resampled.index[0].date()
    end_date       = df_resampled.index[-1].date()
    range_day_cnt  = (end_date - start_date).days
    range_week_cnt = int(range_day_cnt / 7)

    print(f"{rule:6} | {len(df_resampled):>6} bars | "
          f"{start_date} to {end_date} | "
          f"Price range: {df_resampled['low'].min():.0f} -- {df_resampled['high'].max():.0f} | "
          f"{range_day_cnt} days | "
          f"{range_week_cnt} weeks")

    return df_resampled

# -- OPTION 3: Databento | API CONNECTION ----------------------
def load_from_databento(days=config.DATA_DAYS, cache_path=None):
    """Pull `days` of MYM 1-minute bars and resample to 5m + 15m.
    Returns (df_1m, df_5m, df_15m, df_1h); each OHLC with a tz-aware UTC index.
    """
    now    = pd.Timestamp.now(tz="UTC")
    cutoff = now - pd.Timedelta(days=days)

    if cache_path and os.path.exists(cache_path):
        cached = pd.read_csv(cache_path)
        cached["datetime"] = pd.to_datetime(cached["datetime"], utc=True)
        cached = cached.set_index("datetime").sort_index()
        last  = cached.index.max()
        fresh = _pull_1m(start=last)                       # only the gap since last cached bar
        df_1m = pd.concat([cached[["open", "high", "low", "close"]], fresh])
        df_1m = df_1m[~df_1m.index.duplicated(keep="last")].sort_index()
    else:
        df_1m = _pull_1m(start=cutoff)

    df_1m = df_1m[df_1m.index >= cutoff]                   # trim to rolling window
    if cache_path:
        df_1m.reset_index().to_csv(cache_path, index=False)

    df_5m  = resample_ohlc(df_1m, "5min")
    df_15m = resample_ohlc(df_1m, "15min")
    df_1h = resample_ohlc(df_1m, "1h")

    print(f"OK Databento data loaded | 1m: {len(df_1m)} | 5m: {len(df_5m)} | 15m: {len(df_15m)} | 1h: {len(df_1h)}")
    return df_1m, df_5m, df_15m, df_1h


def estimate_databento_cost(days=config.DATA_DAYS):
    """$ cost of the `days`-day 1m pull, WITHOUT spending credits. Call before a backfill."""
    start = pd.Timestamp.now(tz="UTC") - pd.Timedelta(days=days)
    return _client().metadata.get_cost(
        dataset=DATASET, symbols=[SYMBOL], schema=SCHEMA, stype_in=STYPE_IN,
        start=start, end=_safe_end(),
    )

print(f"Est. Databento cost (1030 1m): ${estimate_databento_cost():.4f}")
df_1m, df_5m, df_15m, df_1h = load_from_databento(1030)
```

```python
# Cell 2b — Strategy Functions | MARKET CONTEXT  (== signals.py)
# NOTE (1m migration): set_curve reads 1h, classify_trend reads 15m. When you
# move entries to 1m you will re-decide these two timeframes here.
def set_curve(df_1h, lookback=config.ZONE_LOOKBACK_DAYS):
    recent        = df_1h.tail(lookback)
    highest_high  = recent["high"].max()
    lowest_low    = recent["low"].min()
    total_range   = highest_high - lowest_low
    upper_third   = highest_high - (total_range * 0.3333)
    lower_third   = lowest_low   + (total_range * 0.3333)
    current_price = df_1h["close"].iloc[-1]
    if current_price >= upper_third:   return "RETAIL"
    elif current_price <= lower_third: return "WHOLESALE"
    else:                              return "NEUTRAL"

def classify_trend(df_15m, window=config.TREND_WINDOW):
    if len(df_15m) < window * 2: return "SIDEWAYS"
    rh   = df_15m["high"].rolling(window).max()
    rl   = df_15m["low"].rolling(window).min()
    ph   = rh.shift(window)
    pl   = rl.shift(window)
    last = df_15m.index[-1]
    if rh[last] > ph[last] and rl[last] > pl[last]:   return "UPTREND"
    elif rh[last] < ph[last] and rl[last] < pl[last]: return "DOWNTREND"
    else:                                             return "SIDEWAYS"

curve_result  = set_curve(df_1h_ny)
trend_result  = classify_trend(df_15m_ny)

print("set_curve and classify_trend loaded ✅")
print(f"{'─'*62}")
#----------------------------------------------------------------------------
# ALL LINES BELOW ARE FOR BACKTEST ONLY
print(f"{'─'*15} Curve & Trend Metrics {'─'*15}")
print(f"Curve bias           : {curve_result}")
print(f"Trend                : {trend_result}")
```

```python
# Cell 2d — Strategy Functions | ZONE DETECTION  (== signals.py)
ZONE_HEIGHT_K = 0.38
df = df_15m
min_zone_height = ZONE_HEIGHT_K * df["range"].rolling(20).std().iloc[-1]
print(min_zone_height)

import numpy as np

def _count_taps(price_series, level, side):
    """Distinct returns to `level` after the zone formed. side='below' for DBR
    (low <= level), 'above' for RBD (high >= level). Consecutive in-zone bars
    count as ONE tap — price must leave and return for the next."""
    if price_series.size == 0:
        return 0
    in_zone = (price_series <= level) if side == "below" else (price_series >= level)
    prev = np.concatenate(([False], in_zone[:-1]))
    return int(np.sum(in_zone & ~prev))

def detect_zones(df, max_base_candles=None, min_body_pct=None,
                 range_multiplier=None, min_zone_height=None,
                 min_range_ratio=None, max_body_pct=None):
    if max_base_candles is None: max_base_candles = config.MAX_BASE_CANDLES
    if min_body_pct     is None: min_body_pct     = config.MIN_BODY_PCT
    if range_multiplier is None: range_multiplier = config.RANGE_MULTIPLIER
    if min_zone_height  is None: min_zone_height  = config.MIN_ZONE_HEIGHT
    if min_range_ratio  is None: min_range_ratio  = config.MIN_RANGE_RATIO
    if max_body_pct     is None: max_body_pct     = config.MAX_BODY_PCT
    
    zones = []
    df          = df.copy()
    df["_dt"]   = df.index
    bars        = df.reset_index(drop=True)
    n           = len(bars)
    ranges      = bars["range"].to_numpy()
    body_pcts   = bars["body_pct"].to_numpy()
    highs       = bars["high"].to_numpy()
    lows        = bars["low"].to_numpy()
    directions  = bars["direction"].to_numpy()
    dts         = bars["_dt"].to_numpy()

    avg_rng = df["range"].rolling(20).median().iloc[-1]

    is_base = ((ranges != 0) & (ranges >= 1) &
               (body_pcts <= max_body_pct) &
               (ranges <= avg_rng * range_multiplier))

    i = 2
    while i < n - 1:
        if not is_base[i]:
            i += 1; continue
        base_start = base_end = i
        while base_end + 1 < n - 1:
            if is_base[base_end + 1]: base_end += 1
            else: break
        base_count = base_end - base_start + 1
        if base_count > max_base_candles:
            i = base_end + 1; continue
        base_high   = highs[base_start:base_end + 1].max()
        base_low    = lows[base_start:base_end + 1].min()
        base_height = base_high - base_low
        if base_height < min_zone_height:
            i = base_end + 1; continue
        leg_in_idx  = base_start - 1
        leg_out_idx = base_end + 1
        if not (body_pcts[leg_in_idx] >= min_body_pct and
                ranges[leg_in_idx]    >= base_height * min_range_ratio):
            i = base_end + 1; continue
        if not (body_pcts[leg_out_idx] >= min_body_pct and
                ranges[leg_out_idx]    >= base_height * min_range_ratio):
            i = base_end + 1; continue
        lid = directions[leg_in_idx]; lod = directions[leg_out_idx]
        if   lid == "BEAR" and lod == "BULL": zone_type = "DBR"
        elif lid == "BULL" and lod == "BEAR": zone_type = "RBD"
        else:
            i = base_end + 1; continue
        formed_ts = str(dts[base_end])
        
        # count taps: returns to the proximal line after the zone was established
        prox = base_high if zone_type == "DBR" else base_low
        side = "below" if zone_type == "DBR" else "above"
        post = (lows if zone_type == "DBR" else highs)[leg_out_idx + 1:]
        touched = _count_taps(post, prox, side)
        
        zones.append({
            "type"          : zone_type,
            "proximal"      : round(base_high if zone_type == "DBR" else base_low,  1),
            "distal"        : round(base_low  if zone_type == "DBR" else base_high, 1),
            "base_high"     : round(base_high,        1),
            "base_low"      : round(base_low,         1),
            "base_height"   : round(base_height,      1),
            "base_count"    : base_count,
            "leg_in_range"  : round(ranges[leg_in_idx],  1),
            "leg_out_range" : round(ranges[leg_out_idx], 1),
            "formed_at"     : formed_ts,
            "touched"       : touched,
            "active"        : True,
        })
        i = base_end + 2
    return zones

print("detect_zones loaded")
```

```python
# Cell 2e — Strategy Functions | SCORING  (== signals.py)
def score_zone(zone, trend, curve_bias, all_zones):
    score = 0
    move_ratio = zone["leg_out_range"] / zone["base_height"]
    score += 2 if move_ratio >= 2 else 1 if move_ratio >= 1.5 else 0
    score += 1 if zone["base_count"] <= 3 else 0.5 if zone["base_count"] <= 6 else 0
    score += 2 if zone["touched"] == 0 else 1 if zone["touched"] == 1 else 0
    if (zone["type"] == "DBR" and trend == "UPTREND") or \
       (zone["type"] == "RBD" and trend == "DOWNTREND"):
        score += 1
    if (zone["type"] == "DBR" and curve_bias == "WHOLESALE") or \
       (zone["type"] == "RBD" and curve_bias == "RETAIL"):
        score += 2
    elif curve_bias == "NEUTRAL":
        score += 1
    opposing = [z for z in all_zones if z["type"] != zone["type"] and z["active"]]
    if zone["type"] == "DBR":
        above = [z for z in opposing if z["proximal"] > zone["proximal"]]
        if above:
            target = min(above, key=lambda z: z["proximal"])["proximal"]
            risk   = abs(zone["proximal"] - zone["distal"])
            rr     = abs(target - zone["proximal"]) / risk if risk > 0 else 0
            score += 2 if rr >= 5 else 1 if rr >= 3 else 0
    else:
        below = [z for z in opposing if z["proximal"] < zone["proximal"]]
        if below:
            target = max(below, key=lambda z: z["proximal"])["proximal"]
            risk   = abs(zone["proximal"] - zone["distal"])
            rr     = abs(zone["proximal"] - target) / risk if risk > 0 else 0
            score += 2 if rr >= 5 else 1 if rr >= 3 else 0
    return round(score, 1)

print("score_zone loaded")
```

```python
# Cell 2f — Strategy Functions | ORDER CALCULATION  (== signals.py)
# Cell-8 zone-height value set to 13 (detect_zones default), per request.
# NOTE: this diagnostic references df_15m_ny; run the cell that builds it first.
if "df_15m_ny" in globals():
    stop_buffer = int(df_15m_ny["range"].rolling(20).median().iloc[-1] / 2)
    stop_buffer = min(stop_buffer, int(min_zone_height * 0.7))
    print(f"stop_buffer : {stop_buffer} pts  (cap {int(min_zone_height * 0.7)} pts | "
          f"capped: {stop_buffer == int(min_zone_height * 0.7)})")
else:
    stop_buffer = config.STOP_BUFFER
    print(f"df_1m5_ny not built yet -> stop_buffer defaults to config.STOP_BUFFER={stop_buffer}")

def calculate_sets(zone, all_zones, account_size=ACCOUNT_SIZE,
                   risk_pct=RISK_PCT, tick_value=TICK_VALUE,
                   stop_buffer=config.STOP_BUFFER, t2_cap=T2_CAP,min_zone_height=min_zone_height):
    if zone["type"] == "DBR":
        entry = zone["base_high"]; stop = zone["base_low"] - stop_buffer
    else:
        entry = zone["base_low"];  stop = zone["base_high"] + stop_buffer
    stop_points = abs(entry - stop)
    if stop_points == 0: return None
    direction = 1 if zone["type"] == "DBR" else -1
    max_risk_dollars  = account_size * risk_pct
    risk_per_contract = stop_points * tick_value
    contracts = int(max_risk_dollars / risk_per_contract) if risk_per_contract > 0 else 0
    if contracts < 1: return None
    if zone["type"] == "DBR":
        opposing = [z for z in all_zones if z["type"] == "RBD" and z["proximal"] > entry and z["active"]]
        raw_target = min(opposing, key=lambda z: z["proximal"])["proximal"] if opposing else None
    else:
        opposing = [z for z in all_zones if z["type"] == "DBR" and z["proximal"] < entry and z["active"]]
        raw_target = max(opposing, key=lambda z: z["proximal"])["proximal"] if opposing else None
    if raw_target is None: return None
    raw_target_pts = abs(raw_target - entry)
    rr_raw = round(raw_target_pts / stop_points, 2)
    if rr_raw < 0.5:
        return None
    elif rr_raw <= 1.1:
        t1 = raw_target 
        t2 = None
        t1_points = raw_target_pts
        t2_points = None
        rr_t1 = rr_raw 
        rr_t2 = None
        contracts_t1 = contracts 
        contracts_t2 = None
    elif rr_raw <= 2.0:
        t1 = round(entry + direction * stop_points * 1.0, 1)
        cap_price = round(entry + direction * stop_points * t2_cap, 1)
        t2 = round(min(raw_target, cap_price), 1) if direction == 1 else round(max(raw_target, cap_price), 1)
        t1_points = round(abs(t1 - entry), 1)
        t2_points = round(abs(t2 - entry), 1)
        rr_t1 = round(t1_points / stop_points, 2) 
        rr_t2 = round(t2_points / stop_points, 2)
        contracts_t1 = max(1, contracts // 2)
        contracts_t2 = contracts - contracts_t1
    else:
        t1 = round(entry + direction * stop_points * 1.5, 1)
        cap_price = round(entry + direction * stop_points * t2_cap, 1)
        t2 = round(min(raw_target, cap_price), 1) if direction == 1 else round(max(raw_target, cap_price), 1)
        t1_points = round(abs(t1 - entry), 1)
        t2_points = round(abs(t2 - entry), 1)
        rr_t1 = round(t1_points / stop_points, 2) 
        rr_t2 = round(t2_points / stop_points, 2)
        contracts_t1 = max(1, contracts // 2)
        contracts_t2 = contracts - contracts_t1 if contracts >= 2 else 0
    return {
        "zone_type": zone["type"], 
        "formed_at": zone["formed_at"], 
        "entry": entry, 
        "stop": stop,
        "t1": t1, 
        "t2": t2, 
        "raw_target": raw_target,
        "rr_raw": rr_raw,
        "stop_points": round(stop_points, 1), 
        "t1_points": t1_points, 
        "t2_points": t2_points,
        "rr_t1": rr_t1, 
        "rr_t2": rr_t2, 
        "contracts": contracts,
        "contracts_t1": contracts_t1, 
        "contracts_t2": contracts_t2,
        "risk_per_contract_$": round(risk_per_contract, 2),
        "total_risk_$": round(contracts * risk_per_contract, 2),
        "potential_t1_$": round(contracts_t1 * t1_points * tick_value, 2),
        "potential_t2_$": round(contracts_t2 * t2_points * tick_value, 2) if contracts_t2 is not None else 0
        }
    
print("calculate_sets loaded")
```

```python
# Cell 2g — All Strategy Functions | OUTCOME SIMULATOR
# ── OUTCOME SIMULATOR ─────────────────────────────────────────

def simulate_trade_outcome_dual(entry, stop, t1, t2, zone_type,
                                df_future, max_bars=100):
    CUTOFF_HOUR = 16  # 4:00 PM EST hard cutoff
    t1_hit = False
    for bar_num, (ts, bar) in enumerate(df_future.head(max_bars).iterrows()):
        # ── 1:00 PM EST hard cutoff — close at breakeven ──────
        if ts.tz_convert("US/Eastern").hour >= CUTOFF_HOUR:
            p_t1 = (round(t1-entry,1) if zone_type=="DBR" else round(entry-t1,1)) if t1_hit else 0
            result = "WIN_T1_BE" if t1_hit else "BREAKEVEN"
            return {"result": result, "t1_hit": t1_hit, "t2_hit": False,
                    "bars_held": bar_num+1, "pnl_t1_pts": p_t1,
                    "pnl_t2_pts": 0, "exit_time": str(ts)}
        if zone_type == "DBR":
            # Full loss — stop hit before T1
            if not t1_hit and bar["low"] <= stop:
                return {"result": "LOSS", "t1_hit": False, "t2_hit": False,
                        "bars_held": bar_num+1, "pnl_t1_pts": round(stop-entry,1),
                        "pnl_t2_pts": 0, "exit_time": str(ts)}
            # T1 hit — partial profit, stop moves to breakeven
            if not t1_hit and bar["high"] >= t1:
                t1_hit = True
                stop   = entry  # breakeven
                if t2 is None:
                    return {"result": "WIN_T1", "t1_hit": True, "t2_hit": False,
                            "bars_held": bar_num+1, "pnl_t1_pts": round(t1-entry,1),
                            "pnl_t2_pts": 0, "exit_time": str(ts)}
            # T2 hit — full win
            if t1_hit and t2 and bar["high"] >= t2:
                return {"result": "WIN_T1_T2", "t1_hit": True, "t2_hit": True,
                        "bars_held": bar_num+1, "pnl_t1_pts": round(t1-entry,1),
                        "pnl_t2_pts": round(t2-entry,1), "exit_time": str(ts)}
            # Stopped at breakeven after T1
            if t1_hit and bar["low"] <= entry:
                return {"result": "WIN_T1_BE", "t1_hit": True, "t2_hit": False,
                        "bars_held": bar_num+1, "pnl_t1_pts": round(t1-entry,1),
                        "pnl_t2_pts": 0, "exit_time": str(ts)}
        else:  # RBD short
            # Full loss — stop hit before T1
            if not t1_hit and bar["high"] >= stop:
                return {"result": "LOSS", "t1_hit": False, "t2_hit": False,
                        "bars_held": bar_num+1, "pnl_t1_pts": round(entry-stop,1),
                        "pnl_t2_pts": 0, "exit_time": str(ts)}
            # T1 hit — partial profit, stop moves to breakeven
            if not t1_hit and bar["low"] <= t1:
                t1_hit = True
                stop   = entry  # breakeven
                if t2 is None:
                    return {"result": "WIN_T1", "t1_hit": True, "t2_hit": False,
                            "bars_held": bar_num+1, "pnl_t1_pts": round(entry-t1,1),
                            "pnl_t2_pts": 0, "exit_time": str(ts)}
            # T2 hit — full win
            if t1_hit and t2 and bar["low"] <= t2:
                return {"result": "WIN_T1_T2", "t1_hit": True, "t2_hit": True,
                        "bars_held": bar_num+1, "pnl_t1_pts": round(entry-t1,1),
                        "pnl_t2_pts": round(entry-t2,1), "exit_time": str(ts)}
            # Stopped at breakeven after T1
            if t1_hit and bar["high"] >= entry:
                return {"result": "WIN_T1_BE", "t1_hit": True, "t2_hit": False,
                        "bars_held": bar_num+1, "pnl_t1_pts": round(entry-t1,1),
                        "pnl_t2_pts": 0, "exit_time": str(ts)}
    # max_bars reached — WIN_T1_BE if T1 already hit, otherwise EXPIRED
    p_t1 = (round(t1-entry,1) if zone_type=="DBR" else round(entry-t1,1)) if t1_hit else 0
    last_bar_ts = str(df_future.iloc[min(max_bars, len(df_future)) - 1].name)
    return {"result": "WIN_T1_BE" if t1_hit else "EXPIRED", "t1_hit": t1_hit, "t2_hit": False,
            "bars_held": max_bars, "pnl_t1_pts": p_t1,
            "pnl_t2_pts": 0,
            "exit_time": last_bar_ts}

print("simulate_trade_outcome_dual loaded ✅")
```

```python
# Cell 3 — Run Full Backtest

# Fix index types first
df_1m_ny.index  = pd.to_datetime(df_1m_ny.index).tz_convert("US/Eastern")
df_15m_ny.index = pd.to_datetime(df_15m_ny.index).tz_convert("US/Eastern")
df_1h_ny.index  = pd.to_datetime(df_1h_ny.index).tz_convert("US/Eastern")

min_score       = MIN_SCORE
max_body_pct    = MAX_BODY_PCT
ENTRY_HOUR_MIN  = 7     # earliest ET hour an entry may trigger
ENTRY_HOUR_MAX  = 16    # exclusive — no entries at/after this hour

def run_backtest(df_1m_ny, df_15m_ny, df_1h_ny,
                 account_size=ACCOUNT_SIZE, min_score=min_score,
                 risk_pct=RISK_PCT, tick_value=TICK_VALUE):

    trade_log    = []
    used_zones   = set()
    trading_days = df_15m_ny.index.normalize().unique()
    day_1m_norm  = df_1m_ny.index.normalize()
    day_15m_norm = df_15m_ny.index.normalize()
    day_1h_norm  = df_1h_ny.index.normalize()
    cutoff_hour  = 16
    days_run     = 0
    last_checkpoint_trades = 0

    # ── disposition counters (whole-run totals) ──
    skipped_score = skipped_taps = skipped_sets = skipped_bracket = 0
    skipped_band = no_touch = past_cutoff = 0          # <-- skipped_band added

    print(f"Trading days in full dataset: {len(trading_days)}")

    for day in trading_days:
        lookback_start = day - pd.Timedelta(days=ZONE_LOOKBACK_DAYS)
        zone_lookback  = day - pd.Timedelta(days=TREND_WINDOW)

        bars_15m_zones = df_15m_ny[(day_15m_norm >= zone_lookback) & (day_15m_norm <= day)].copy()
        bars_15m       = df_15m_ny[(day_15m_norm >= lookback_start) & (day_15m_norm <= day)].copy()
        bars_1h        = df_1h_ny[(day_1h_norm  >= lookback_start) & (day_1h_norm  <= day)].copy()

        if len(bars_15m_zones) < 50 or len(bars_15m) < 20 or len(bars_1h) < 50:
            continue

        days_run += 1
        if days_run % 50 == 0:
            added = len(trade_log) - last_checkpoint_trades
            print(f"  Day {days_run} | {str(day.date())} | "
                  f"trades: {len(trade_log)} (+{added} in last 50d)")
            last_checkpoint_trades = len(trade_log)

        try:
            curve_bias = set_curve(bars_1h)
            trend      = classify_trend(bars_15m)
        except Exception:
            continue

        # ── σ-band, computed AS OF `day` (mirrors live's single-run band) ──
        range_std     = bars_15m["range"].rolling(config.ENTRY_SIGMA_WINDOW).std().iloc[-1]
        current_price = bars_15m["close"].iloc[-1]
        entry_band    = config.ENTRY_SIGMA * range_std

        zones = detect_zones(bars_15m_zones)
        if not zones:
            continue

        for zone in zones:
            zone_id = f"{zone['type']}_{zone['proximal']}_{zone['formed_at']}"
            if zone_id in used_zones:
                continue

            # ── pre-entry disposition gates (order matches run_signal_check) ──
            score = score_zone(zone, trend, curve_bias, zones)
            if score < min_score:
                skipped_score += 1;  continue
            if zone["touched"] >= config.MAX_TAPS:
                skipped_taps += 1;   continue

            sets = calculate_sets(zone, zones, account_size=account_size,
                                  risk_pct=risk_pct, tick_value=tick_value)
            if sets is None:
                skipped_sets += 1;   continue

            # ── σ-band gate (NEW — mirrors live) ──
            dist = abs(sets["entry"] - current_price)
            if pd.notna(range_std) and range_std > 0 and dist > entry_band:
                skipped_band += 1;   continue

            # bracket structural validity
            e, s, t1, t2 = sets["entry"], sets["stop"], sets["t1"], sets["t2"]
            if zone["type"] == "DBR":
                bracket_ok = (s < e < t1) and (t2 is None or e < t2)
            else:
                bracket_ok = (t1 < e < s) and (t2 is None or t2 < e)
            if not bracket_ok:
                skipped_bracket += 1
                print(f"  !! MALFORMED BRACKET {zone['type']} @ {e} "
                      f"stop={s} t1={t1} t2={t2} (formed {zone['formed_at']})")
                continue

            # ── entry search ──
            signal_time = pd.Timestamp(zone["formed_at"])
            entry_price = sets["entry"]
            zone_type   = zone["type"]

            candidate_bars = df_1m_ny[(df_1m_ny.index > signal_time) &
                                      (day_1m_norm <= day)].copy()
            candidate_bars = candidate_bars[candidate_bars.index.hour >= ENTRY_HOUR_MIN]
            if len(candidate_bars) == 0:
                no_touch += 1;  continue

            if zone_type == "DBR":
                touch_mask = candidate_bars["close"] >= entry_price
            else:
                touch_mask = candidate_bars["close"] <= entry_price
            touch_bars = candidate_bars[touch_mask]
            if len(touch_bars) == 0:
                no_touch += 1;  continue

            entry_bar  = touch_bars.index[0]
            entry_day  = entry_bar.normalize()
            entry_time = str(entry_bar)

            if entry_bar.hour >= cutoff_hour:
                past_cutoff += 1;  continue

            future_bars = candidate_bars[(candidate_bars.index >= entry_bar) &
                                         (candidate_bars.index.normalize() == entry_day)].copy()
            if len(future_bars) == 0:
                no_touch += 1;  continue

            outcome = simulate_trade_outcome_dual(
                sets["entry"], sets["stop"], sets["t1"], sets["t2"],
                zone["type"], future_bars)

            pnl_t1    = outcome["pnl_t1_pts"] * tick_value * sets["contracts_t1"]
            pnl_t2    = outcome["pnl_t2_pts"] * tick_value * sets["contracts_t2"] \
                        if sets["contracts_t2"] is not None else 0
            total_pnl = round(pnl_t1 + pnl_t2, 2)

            trade_log.append({
                "day": str(entry_day.date()), "formed_at": zone["formed_at"],
                "entry_time": entry_time, "zone_type": zone["type"], "score": score,
                "trend": trend, "curve": curve_bias,
                "entry": sets["entry"], "stop": sets["stop"], "t1": sets["t1"], "t2": sets["t2"],
                "raw_target": sets["raw_target"], "rr_raw": sets["rr_raw"],
                "stop_pts": sets["stop_points"], "t1_pts": sets["t1_points"], "t2_pts": sets["t2_points"],
                "rr_t1": sets["rr_t1"], "rr_t2": sets["rr_t2"],
                "contracts": sets["contracts"], "contracts_t1": sets["contracts_t1"],
                "contracts_t2": sets["contracts_t2"],
                "result": outcome["result"], "t1_hit": outcome["t1_hit"], "t2_hit": outcome["t2_hit"],
                "bars_held": outcome["bars_held"],
                "pnl_t1_pts": outcome["pnl_t1_pts"], "pnl_t2_pts": outcome["pnl_t2_pts"],
                "pnl_dollars": total_pnl, "exit_time": outcome["exit_time"],
            })
            used_zones.add(zone_id)

    print(f"Days passed warm-up gate: {days_run}")
    if skipped_bracket:
        print(f"!!!! {skipped_bracket} MALFORMED BRACKET(S) — CHECK calculate_sets !!!!")
    print(f"Disposition: {len(trade_log)} traded | {skipped_score} score | "
          f"{skipped_taps} taps | {skipped_sets} no-sets | {skipped_band} band | "
          f"{skipped_bracket} bracket | {no_touch} no-touch | {past_cutoff} past-cutoff")
    return trade_log

from pyinstrument import Profiler

USE_PARTIAL = False # flip to False for a full run

if USE_PARTIAL:
    cutoff_day = df_15m_ny.index.normalize().unique()[-800]
    df_1m_prof  = df_1m_ny[df_1m_ny.index.normalize()  <= cutoff_day]
    df_15m_prof  = df_15m_ny[df_15m_ny.index.normalize()  <= cutoff_day]
    df_1h_prof = df_1h_ny[df_1h_ny.index.normalize() <= cutoff_day]
else:
    df_1m_prof  = df_1m_ny
    df_15m_prof  = df_15m_ny
    df_1h_prof = df_1h_ny


print("Running backtest (1m expected run time: 2.5M rows raw data)...")
profiler = Profiler()
profiler.start()
try:
    trade_log_ETRADEV3_QA = run_backtest(df_1m_prof, df_15m_prof, df_1h_prof,
                                    min_score=min_score)
except KeyboardInterrupt:
    trade_log_ETRADEV3_QA = []
    print("Interrupted — showing partial profile from the time elapsed so far.\n")
finally:
    profiler.stop()
    profiler.print(color=False, show_all=False)
    profiler.write_html("run_backtest_profile.html")
    print("\nInteractive report written to run_backtest_profile.html (open in a browser)")

print(f"\nTotal trades logged: {len(trade_log_ETRADEV3_QA)}")

```

```python
# Cell 4 — Run Performance Analysis and Exam Scorecard (Dual-Target)
import numpy as np
import os
from dotenv import load_dotenv
NB_BASE = "/Users/efiles/prod/etrade"
load_dotenv(os.path.join(NB_BASE, ".env"))
sys.path.insert(0, NB_BASE)          # where this env's config.py lives
import config
results = pd.DataFrame(trade_log_ETRADEV3_QA) if trade_log_ETRADEV3_QA else pd.DataFrame()
account_size=ACCOUNT_SIZE 

if len(results) == 0:
    print("No trades at min_score=7. Check strategy parameters.")
else:
    # ── RESULT CLASSIFICATION ─────────────────────────────────────
    # Wins = any result where T1 was hit (partial or full)
    # Losses = stop hit before T1
    full_wins  = results[results["result"] == "WIN_T1_T2"]
    part_wins  = results[results["result"] == "WIN_T1"]
    be_wins    = results[results["result"] == "WIN_T1_BE"]
    losses     = results[results["result"] == "LOSS"]
    expired    = results[results["result"] == "EXPIRED"]

    all_wins   = results[results["result"].isin(["WIN_T1_T2", "WIN_T1", "WIN_T1_BE"])]

    total      = len(results)
    win_rate   = len(all_wins) / total * 100

    # ── CORE METRICS ──────────────────────────────────────────────
    gross_wins    = all_wins["pnl_dollars"].sum()
    gross_losses  = abs(losses["pnl_dollars"].sum())
    net_pnl       = results["pnl_dollars"].sum()
    avg_win       = all_wins["pnl_dollars"].mean() if len(all_wins) > 0 else 0
    avg_loss      = losses["pnl_dollars"].mean()   if len(losses)   > 0 else 0
    profit_factor = gross_wins / gross_losses       if gross_losses  > 0 else float("inf")
    expectancy    = (win_rate/100 * avg_win) + ((1 - win_rate/100) * avg_loss)

    returns    = results["pnl_dollars"].values
    mean_ret   = np.mean(returns)
    std_ret    = np.std(returns)
    sharpe     = mean_ret / std_ret if std_ret > 0 else 0

    neg_returns = returns[returns < 0]
    down_std    = np.std(neg_returns) if len(neg_returns) > 0 else 0
    sortino     = mean_ret / down_std if down_std > 0 else 0

    results["cum_pnl"] = results["pnl_dollars"].cumsum()
    results["peak"]    = results["cum_pnl"].cummax()
    results["dd"]      = results["cum_pnl"] - results["peak"]
    max_dd_pct         = results["dd"].min() / account_size * 100

    # T1 and T2 hit rates
    t1_hit_rate = results["t1_hit"].mean() * 100 if "t1_hit" in results.columns else 0
    t2_hit_rate = results["t2_hit"].mean() * 100 if "t2_hit" in results.columns else 0

    avg_rr_t1 = results["rr_t1"].mean() if "rr_t1" in results.columns else 0
    avg_rr_t2 = results["rr_t2"].mean() if "rr_t2" in results.columns else 0

    # ── PRINT REPORT ──────────────────────────────────────────────
    print("=" * 60)
    print("  OTA STRATEGY — DUAL-TARGET PERFORMANCE REPORT")
    print("=" * 60)
    print(f"  Total Trades       : {total}")
    print(f"  Full Wins (T1+T2)  : {len(full_wins)}")
    print(f"  Partial Wins (T1)  : {len(part_wins)}")
    print(f"  Breakeven (T1+BE)  : {len(be_wins)}")
    print(f"  Losses             : {len(losses)}")
    print(f"  Expired            : {len(expired)}")
    print(f"  ─────────────────────────────────────────────")
    print(f"  Win Rate           : {win_rate:.1f}%  (any T1 hit)")
    print(f"  T1 Hit Rate        : {t1_hit_rate:.1f}%")
    print(f"  T2 Hit Rate        : {t2_hit_rate:.1f}%")
    print(f"  ─────────────────────────────────────────────")
    print(f"  Avg Win            : ${avg_win:.2f}")
    print(f"  Avg Loss           : ${avg_loss:.2f}")
    print(f"  Avg R:R T1         : {avg_rr_t1:.2f}:1")
    print(f"  Avg R:R T2         : {avg_rr_t2:.2f}:1")
    print(f"  ─────────────────────────────────────────────")
    print(f"  Gross Wins         : ${gross_wins:.2f}")
    print(f"  Gross Losses       : ${gross_losses:.2f}")
    print(f"  Net P&L            : ${net_pnl:.2f}")
    print(f"  Profit Factor      : {profit_factor:.2f}")
    print(f"  Expectancy/Trade   : ${expectancy:.2f}")
    print(f"  Sharpe Ratio       : {sharpe:.2f}")
    print(f"  Sortino Ratio      : {sortino:.2f}")
    print(f"  Max Drawdown       : {max_dd_pct:.2f}%")
    print()

    # ── RESULT BREAKDOWN ──────────────────────────────────────────
    print("  RESULT BREAKDOWN")
    print(f"  {'─'*52}")
    for r, label in [
        ("WIN_T1_T2", "Full Win  (T1+T2 hit)"),
        ("WIN_T1",    "Partial   (T1 only)  "),
        ("WIN_T1_BE", "Breakeven (T1+BE)    "),
        ("LOSS",      "Loss      (stop hit) "),
        ("EXPIRED",   "Expired   (max bars) "),
    ]:
        subset = results[results["result"] == r]
        if len(subset) == 0: continue
        net = subset["pnl_dollars"].sum()
        avg = subset["pnl_dollars"].mean()
        print(f"  {label} : {len(subset):>3} trades | "
              f"Net: ${net:>8.2f} | Avg: ${avg:>7.2f}")
    print()

    # ── EXAM SCORECARD ────────────────────────────────────────────
    print("  MID-TERM EXAM SCORECARD")
    print(f"  {'─'*52}")
    checks = [
        ("Total Trades ≥ 30",    total >= 30,              str(total)),
        ("Win Rate ≥ 35%",        win_rate >= 35,           f"{win_rate:.1f}%"),
        ("Profit Factor ≥ 1.3",  profit_factor >= 1.3,     f"{profit_factor:.2f}"),
        ("Expectancy > $0",       expectancy > 0,           f"${expectancy:.2f}"),
        ("Max Drawdown < 20%",    abs(max_dd_pct) < 20,     f"{max_dd_pct:.2f}%"),
        ("Sharpe ≥ 0.5",          sharpe >= 0.5,            f"{sharpe:.2f}"),
        ("Sortino ≥ 2",          sortino >= 2,            f"{sortino:.1f}"),
        
    ]
    passed = 0
    for label, condition, value in checks:
        icon = "✅" if condition else "❌"
        if condition: passed += 1
        print(f"  {icon} {label:<30} {value}")

    print(f"\n  Score: {passed}/{len(checks)-1} checks passed")
    if passed == len(checks):
        verdict = "🟢 PASSED — Edge confirmed. Phase 2 unlocked."
    elif passed >= 4:
        verdict = "🟡 CLOSE — Strong result. See details below."
    else:
        verdict = "🔴 NEEDS MORE DATA OR PARAMETER REVIEW"
    print(f"  VERDICT: {verdict}")
    print("=" * 60)


    # ── SAVE ──────────────────────────────────────────────────────
    os.makedirs(NB_BASE, exist_ok=True)
    results.to_csv(f"{NB_BASE}/data/trade_log_ETRADEV3_QA.csv", index=False)
    print(f"\nTrade log saved → {NB_BASE}/data/trade_log_ETRADEV3_QA.csv")
    
    
    
```

```python
# ── QA: Freshness re-grade — before vs after touched becomes real ──
# Scores every detected zone two ways: with touched=0 (old behavior) and with
# the real touched value (new detect_zones). Shows how many zones the freshness
# change re-grades, how many that alone drops below MIN_SCORE, and how much the
# explicit MAX_TAPS filter adds on top.
#
# Requires: the UPDATED detect_zones (real touched) is already defined.

import pandas as pd

MIN_SCORE = config.MIN_SCORE
MAX_TAPS  = config.MAX_TAPS
SAMPLE_EVERY = 1          # raise to 5/10 to subsample days if this is slow

records = []
trading_days = df_15m_ny.index.normalize().unique()
day_15m_norm = df_15m_ny.index.normalize()
day_1h_norm  = df_1h_ny.index.normalize()

for k, day in enumerate(trading_days):
    if k % SAMPLE_EVERY != 0:
        continue
    lookback_start = day - pd.Timedelta(days=60)
    zone_lookback  = day - pd.Timedelta(days=120)

    bars_15m_zones = df_15m_ny[(day_15m_norm >= zone_lookback) & (day_15m_norm <= day)].copy()
    bars_15m       = df_15m_ny[(day_15m_norm >= lookback_start) & (day_15m_norm <= day)].copy()
    bars_1h        = df_1h_ny[(day_1h_norm  >= lookback_start) & (day_1h_norm  <= day)].copy()
    if len(bars_15m_zones) < 50 or len(bars_15m) < 20 or len(bars_1h) < 50:
        continue

    try:
        curve_bias = set_curve(bars_1h)
        trend      = classify_trend(bars_15m)
    except Exception:
        continue

    zones = detect_zones(bars_15m_zones)
    for z in zones:
        real_touched = z["touched"]
        try:
            score_after = score_zone(z, trend, curve_bias, zones)   # real touched
            z_tmp = dict(z); z_tmp["touched"] = 0                   # old behavior
            score_before = score_zone(z_tmp, trend, curve_bias, zones)
        except Exception:
            continue
        records.append({
            "day": str(day.date()), "type": z["type"], "touched": real_touched,
            "score_before": score_before, "score_after": score_after,
            "delta": round(score_after - score_before, 1),
        })

g = pd.DataFrame(records)
print(f"zones evaluated: {len(g)}")
print(f"\ntouched distribution:\n{g['touched'].value_counts().sort_index().to_string()}")

regraded   = g[g["delta"] != 0]
passed_b   = g["score_before"] >= MIN_SCORE
passed_a   = g["score_after"]  >= MIN_SCORE
dropped_by_score = g[passed_b & ~passed_a]                 # re-grade alone removes these
removed_by_filter = g[g["touched"] >= MAX_TAPS]
filter_unique = removed_by_filter[removed_by_filter["score_after"] >= MIN_SCORE]

print(f"\nzones re-graded (score changed): {len(regraded)} "
      f"({len(regraded)/len(g)*100:.1f}%)")
print(f"passed MIN_SCORE before: {passed_b.sum()}  |  after re-grade: {passed_a.sum()}")
print(f"dropped below MIN_SCORE by the re-grade ALONE: {len(dropped_by_score)}")
print(f"removed by MAX_TAPS>={MAX_TAPS} filter: {len(removed_by_filter)}")
print(f"  -> of those, already below MIN_SCORE from re-grade: "
      f"{len(removed_by_filter) - len(filter_unique)}")
print(f"  -> filter's UNIQUE contribution (passed score but tapped-out): {len(filter_unique)}")

# distribution shift
print("\nscore distribution before -> after:")
comp = pd.DataFrame({
    "before": g["score_before"].value_counts().sort_index(),
    "after":  g["score_after"].value_counts().sort_index(),
}).fillna(0).astype(int)
print(comp.to_string())
```

```python
# Simulate prop firm account days to pass

import numpy as np, pandas as pd

# Build daily P&L at $250 risk
r = results.copy()
TICK = 0.50
r["risk_dollars"] = r["stop_pts"] * r["contracts"] * TICK
r["R"]            = r["pnl_dollars"] / r["risk_dollars"]
r["pnl_250"]      = r["R"] * 200
daily_all = r.groupby("day")["pnl_250"].sum().sort_index().values

TARGET, TRAIL, START = 3000.0, 2000.0, 50000.0
N_SIMS, MAX_DAYS = 10000, 30
RECENT_DAYS = 30

def simulate(pool, label):
    rng = np.random.default_rng(42)
    npass = blow = timeout = 0
    days_to_pass = []
    for _ in range(N_SIMS):
        bal, hwm, trail = START, START, START - TRAIL
        for d in range(1, MAX_DAYS + 1):
            bal += rng.choice(pool)
            if bal > hwm:
                hwm, trail = bal, bal - TRAIL
            if bal <= trail:
                blow += 1; break
            if bal - START >= TARGET:
                npass += 1; days_to_pass.append(d); break
        else:
            timeout += 1
    dd = np.array(days_to_pass) if days_to_pass else np.array([0])
    print(f"\n── {label} (pool = {len(pool)} days) ──")
    print(f"  Pass rate      : {npass/N_SIMS*100:5.1f}%")
    print(f"  Blow-up rate   : {blow/N_SIMS*100:5.1f}%")
    print(f"  Timeout (>120d): {timeout/N_SIMS*100:5.1f}%")
    print(f"  Median days    : {np.median(dd):.0f}   (IQR {np.percentile(dd,25):.0f}–{np.percentile(dd,75):.0f})")

simulate(daily_all,               "FULL HISTORY")
simulate(daily_all[-RECENT_DAYS:], "RECENT 120 DAYS")
```

```python
# QA 11 - Zone detection diagnostic (detect_zones filter check)
bars        = df.reset_index(drop=True)
bars["_dt"] = bars.index
highs       = bars["high"].to_numpy()
lows        = bars["low"].to_numpy()
avg_rng = df["range"].rolling(20).median().iloc[-1]

n = len(bars)
base_candidates     = 0
base_too_long       = 0
height_too_small    = 0
leg_in_fail         = 0
leg_out_fail        = 0
direction_fail      = 0
zones_passed        = 0

i = 2
while i < n - 1:
    row = bars.iloc[i]
    if not is_base_candle(row, avg_rng, range_multiplier=0.53):
        i += 1; continue

    base_start = base_end = i
    while base_end + 1 < n - 1:
        if is_base_candle(bars.iloc[base_end + 1], avg_rng, range_multiplier=0.53):
            base_end += 1
        else:
            break

    base_candidates += 1
    base_count = base_end - base_start + 1

    if base_count > 6:
        base_too_long += 1
        i = base_end + 1; continue

    base_high = highs[base_start:base_end + 1].max()
    base_low  = lows[base_start:base_end + 1].min()
    base_height = base_high - base_low

    if base_height < 13:
        height_too_small += 1
        i = base_end + 1; continue

    leg_in  = bars.iloc[base_start - 1]
    leg_out = bars.iloc[base_end  + 1]

    if not is_valid_leg(leg_in, base_height, min_range_ratio=0.639):
        leg_in_fail += 1
        i = base_end + 1; continue

    if not is_valid_leg(leg_out, base_height, min_range_ratio=0.639):
        leg_out_fail += 1
        i = base_end + 1; continue

    lid = leg_in["direction"]
    lod = leg_out["direction"]
    if not ((lid == "BEAR" and lod == "BULL") or (lid == "BULL" and lod == "BEAR")):
        direction_fail += 1
        i = base_end + 1; continue

    zones_passed += 1
    i = base_end + 2

print(f"avg_rng              : {avg_rng:.1f} pts")
print(f"Base candle ceiling  : {avg_rng * 0.53:.1f} pts")
print(f"Leg candle floor     : {avg_rng * 0.81:.1f} pts  (× base_height)")
print(f"{'─'*45}")
print(f"Base candidates      : {base_candidates}")
print(f"  Failed — too long  : {base_too_long}")
print(f"  Failed — too small : {height_too_small}")
print(f"  Failed — leg_in    : {leg_in_fail}")
print(f"  Failed — leg_out   : {leg_out_fail}")
print(f"  Failed — direction : {direction_fail}")
print(f"Zones passed         : {zones_passed}")
```
